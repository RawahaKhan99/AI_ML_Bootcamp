Difference between Numpy & Pandas

Numpy: 
- Limited/No Support for different file formats; no file reading
- no missing element concept
- Disguised indexing of elements
- No grouping of data


Pandas:
- Does support different formats
- Does contain missing element concept: NaN, NaT
- Indexing appears alongside; Labelled rows and columns
- groupby, pivot_tables; effective for data manipulation
