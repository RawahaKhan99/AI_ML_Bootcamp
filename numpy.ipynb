Time Difference between a Standard Python list and a NumPy Array

```python
import time
import numpy as np

n = 1_000

# -------------------------
# Python List
# -------------------------

py_list = list(range(n))

start = time.time()

for i in range(len(py_list)):
    py_list[i] = py_list[i] * 2

end = time.time()

print("Time taken for Python list:", end - start, "seconds")
print(py_list)


# -------------------------
# NumPy Array
# -------------------------

np_array = np.arange(n)

start = time.time()

np_array = np_array * 2

end = time.time()

print("Time taken for NumPy:", end - start, "seconds")
print(np_array)
```
